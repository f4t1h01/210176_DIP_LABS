```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


image = cv2.imread('a.jpg')



# Creating our 3 x 3 kernel
kernel_3x3 = np.ones((3, 3), np.float32) / 9

# We use the cv2.fitler2D to conovlve the kernal with an image
blurred = cv2.filter2D(image, -1, kernel_3x3)


# Creating our 5 x 5 kernel
kernel_5x5 = np.ones((5, 5), np.float32) / 25

blurred2 = cv2.filter2D(image, -1, kernel_5x5)


# Creating our 9 x 9 kernel
kernel_9x9 = np.ones((9, 9), np.float32) / 81

blurred3 = cv2.filter2D(image, -1, kernel_9x9)

plt.figure(figsize=[16,8])
plt.subplot(221);plt.imshow(image[...,::-1]);plt.title("Original")
plt.subplot(222);plt.imshow(blurred[...,::-1]);plt.title("blurred with 3x3 filter")
plt.subplot(223);plt.imshow(blurred2[...,::-1]);plt.title("blurred with 5x5 filter")
plt.subplot(224);plt.imshow(blurred3[...,::-1]);plt.title("blurred with 9x9 filter")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('a.jpg')


gaussian_s1 = cv2.GaussianBlur(image, (7,7), 1)

gaussian_s2 = cv2.GaussianBlur(image, (7,7), 5)

gaussian_s3 = cv2.GaussianBlur(image, (7,7), 15)


plt.figure(figsize=[16,8])
plt.subplot(221);plt.imshow(image[...,::-1]);plt.title("Original")
plt.subplot(222);plt.imshow(gaussian_s1[...,::-1]);plt.title("After gaussian filter with sigma 1")
plt.subplot(223);plt.imshow(gaussian_s2[...,::-1]);plt.title("After gaussian filter with sigma 5")
plt.subplot(224);plt.imshow(gaussian_s3[...,::-1]);plt.title("After gaussian filter with sigma 15")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


image = cv2.imread('a.jpg')


# Creating our 3 x 3 kernel
kernel_3x3 =  np.array([[0,0,0],
                        [0,0.5,0],
                        [0,0,0]])

# We use the cv2.fitler2D to conovlve the kernal with an image
sharped = cv2.filter2D(image, -1, kernel_3x3)

plt.figure(figsize=[16,8])
plt.subplot(121);plt.imshow(image[...,::-1]);plt.title("Original")
plt.subplot(122);plt.imshow(sharped[...,::-1]);plt.title("blurred with 3x3 filter")
```
